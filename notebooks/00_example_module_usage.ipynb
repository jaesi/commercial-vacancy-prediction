## 4. 모델링

`src.modeling` 모듈의 함수들을 사용합니다.

```python
from src.modeling import (
    split_data,
    train_random_forest,
    train_gradient_boosting,
    evaluate_model,
    get_feature_importance,
    RF_PARAM_GRID
)

# 예제 데이터 생성
X = pd.DataFrame(
    np.random.randn(100, 5),
    columns=['feature1', 'feature2', 'feature3', 'feature4', 'feature5']
)
y = pd.Series(np.random.uniform(10, 40, 100))

# 데이터 분할
X_train, X_test, y_train, y_test = split_data(X, y, test_size=0.2)
print(f"학습 데이터 크기: {X_train.shape}")
print(f"테스트 데이터 크기: {X_test.shape}")
```

### Random Forest 모델 학습

```python
# Random Forest 모델 학습
rf_model = train_random_forest(X_train, y_train, params={'n_estimators': 100, 'random_state': 42})

# 모델 평가
results = evaluate_model(rf_model, X_test, y_test)

# 특성 중요도 확인
feature_importance_df = get_feature_importance(rf_model, X.columns)
print("\n특성 중요도:")
print(feature_importance_df)
```
